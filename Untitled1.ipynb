```python
# ============================================
# AI Semester Project: Student AI Impact Analysis
# Author: AI Assistant
# Date: June 2026
# ============================================

# %% [markdown]
# # 📊 AI Impact on Student Academic Performance
#
# ## Project Objectives:
# 1. Analyze how weekly AI usage affects GPA and skill retention
# 2. Identify optimal AI usage patterns across different majors
# 3. Predict student burnout risk using AI usage patterns
# 4. Provide actionable recommendations for students and institutions

# %% [markdown]
# ### 1. Setup and Data Loading

# %%
# Install required libraries (if not already installed)
!pip install -q pandas numpy matplotlib seaborn scikit-learn xgboost imbalanced-learn

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import pearsonr, spearmanr

# Machine Learning imports
import sklearn
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (mean_squared_error, r2_score, accuracy_score,
                             classification_report, confusion_matrix, roc_auc_score)
from xgboost import XGBRegressor, XGBClassifier
from imblearn.over_sampling import SMOTE

import warnings
warnings.filterwarnings('ignore')

# Set visualization style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

# Print versions
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
print(f"Scikit-learn version: {sklearn.__version__}")

# %% [markdown]
# ### 2. Load and Explore the Dataset

# %%
# Load the dataset
from google.colab import files
import io

# Upload the file (you'll need to upload the CSV file when prompted)
print("Please upload your CSV file:")
uploaded = files.upload()

# Get the filename
filename = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[filename]))

# For this project, let's also provide a fallback in case you want to use the data directly
# You can copy the data into a variable or save it as a file

print(f"Dataset loaded successfully! Shape: {df.shape}")

# %%
# Display basic information about the dataset
print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print(f"\nNumber of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")

print("\nColumn names and data types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())

# %%
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())

print("\nBasic Statistics:")
display(df.describe())

# %% [markdown]
# ### 3. Data Cleaning and Feature Engineering

# %%
# Check for duplicates
print(f"Duplicate rows: {df.duplicated().sum()}")

# Create a copy for processing
df_clean = df.copy()

# Encode categorical variables
categorical_columns = ['Major_Category', 'Primary_Use_Case', 'Prompt_Engineering_Skill',
                       'Institutional_Policy', 'Burnout_Risk_Level', 'Paid_Subscription']

label_encoders = {}
for col in categorical_columns:
    le = LabelEncoder()
    df_clean[f'{col}_encoded'] = le.fit_transform(df_clean[col])
    label_encoders[col] = le
    print(f"Encoded {col}: {dict(zip(le.classes_, le.transform(le.classes_)))}")

# Calculate GPA change (improvement or decline)
df_clean['GPA_Change'] = df_clean['Post_Semester_GPA'] - df_clean['Pre_Semester_GPA']
df_clean['GPA_Improvement'] = (df_clean['GPA_Change'] > 0).astype(int)

# Create AI intensity categories
df_clean['AI_Usage_Category'] = pd.cut(df_clean['Weekly_GenAI_Hours'],
                                         bins=[0, 5, 15, 30, float('inf')],
                                         labels=['Low (0-5h)', 'Medium (5-15h)',
                                                'High (15-30h)', 'Extreme (>30h)'])

# Create tool diversity categories
df_clean['Tool_Diversity_Category'] = pd.cut(df_clean['Tool_Diversity'],
                                              bins=[0, 2, 4, 6, float('inf')],
                                              labels=['Limited', 'Moderate', 'High', 'Very High'])

# Create dependency categories
df_clean['Dependency_Category'] = pd.cut(df_clean['Perceived_AI_Dependency'],
                                          bins=[0, 3, 6, 8, float('inf')],
                                          labels=['Low', 'Medium', 'High', 'Very High'])

# Create anxiety categories
df_clean['Anxiety_Category'] = pd.cut(df_clean['Anxiety_Level_During_Exams'],
                                       bins=[0, 3, 6, 8, float('inf')],
                                       labels=['Low', 'Medium', 'High', 'Very High'])

print("\nNew features created:")
print(f"- GPA_Change: {df_clean['GPA_Change'].min():.2f} to {df_clean['GPA_Change'].max():.2f}")
print(f"- AI_Usage_Category: {df_clean['AI_Usage_Category'].value_counts().to_dict()}")

# %% [markdown]
# ### 4. Exploratory Data Analysis (EDA)

# %%
# 4.1 Distribution of AI Usage Hours
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Distribution of Weekly AI Hours
ax1 = axes[0, 0]
df_clean['Weekly_GenAI_Hours'].hist(bins=30, color='skyblue', edgecolor='black', ax=ax1)
ax1.axvline(df_clean['Weekly_GenAI_Hours'].mean(), color='red', linestyle='--',
            label=f"Mean: {df_clean['Weekly_GenAI_Hours'].mean():.1f}h")
ax1.axvline(df_clean['Weekly_GenAI_Hours'].median(), color='green', linestyle='--',
            label=f"Median: {df_clean['Weekly_GenAI_Hours'].median():.1f}h")
ax1.set_xlabel('Weekly AI Hours')
ax1.set_ylabel('Frequency')
ax1.set_title('Distribution of Weekly AI Usage')
ax1.legend()

# Box plot by Major
ax2 = axes[0, 1]
sns.boxplot(data=df_clean, x='Major_Category', y='Weekly_GenAI_Hours', ax=ax2)
ax2.set_xlabel('Major Category')
ax2.set_ylabel('Weekly AI Hours')
ax2.set_title('AI Hours by Major')
ax2.tick_params(axis='x', rotation=45)

# Distribution of GPA Change
ax3 = axes[1, 0]
df_clean['GPA_Change'].hist(bins=30, color='lightgreen', edgecolor='black', ax=ax3)
ax3.axvline(0, color='red', linestyle='--', label='No Change')
ax3.set_xlabel('GPA Change')
ax3.set_ylabel('Frequency')
ax3.set_title('Distribution of GPA Change')
ax3.legend()

# Pie chart of Burnout Risk
ax4 = axes[1, 1]
burnout_counts = df_clean['Burnout_Risk_Level'].value_counts()
colors = ['#ff9999', '#66b3ff', '#99ff99']
ax4.pie(burnout_counts.values, labels=burnout_counts.index, autopct='%1.1f%%',
        colors=colors, startangle=90)
ax4.set_title('Burnout Risk Distribution')

plt.tight_layout()
plt.show()

# %%
# 4.2 Correlation Analysis
# Select numeric columns for correlation
numeric_cols = ['Pre_Semester_GPA', 'Weekly_GenAI_Hours', 'Tool_Diversity',
                'Traditional_Study_Hours', 'Perceived_AI_Dependency',
                'Anxiety_Level_During_Exams', 'Post_Semester_GPA', 'Skill_Retention_Score']

correlation_matrix = df_clean[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
            square=True, fmt='.2f')
plt.title('Correlation Matrix of Key Metrics')
plt.tight_layout()
plt.show()

# Print key correlations
print("\n🔍 KEY CORRELATIONS:")
print(f"AI Hours vs Post GPA: {correlation_matrix.loc['Weekly_GenAI_Hours', 'Post_Semester_GPA']:.3f}")
print(f"AI Hours vs Skill Retention: {correlation_matrix.loc['Weekly_GenAI_Hours', 'Skill_Retention_Score']:.3f}")
print(f"AI Dependency vs Anxiety: {correlation_matrix.loc['Perceived_AI_Dependency', 'Anxiety_Level_During_Exams']:.3f}")
print(f"Study Hours vs Post GPA: {correlation_matrix.loc['Traditional_Study_Hours', 'Post_Semester_GPA']:.3f}")

# %%
# 4.3 AI Usage vs GPA by Major
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Scatter plot with trend line
ax1 = axes[0]
majors = df_clean['Major_Category'].unique()
colors_map = plt.cm.Set3(np.linspace(0, 1, len(majors)))

for major, color in zip(majors, colors_map):
    subset = df_clean[df_clean['Major_Category'] == major]
    ax1.scatter(subset['Weekly_GenAI_Hours'], subset['Post_Semester_GPA'],
                alpha=0.6, label=major, color=color, s=30)

    # Add trend line
    z = np.polyfit(subset['Weekly_GenAI_Hours'], subset['Post_Semester_GPA'], 1)
    p = np.poly1d(z)
    ax1.plot(np.sort(subset['Weekly_GenAI_Hours']),
             p(np.sort(subset['Weekly_GenAI_Hours'])),
             color=color, linewidth=1.5)

ax1.set_xlabel('Weekly AI Hours')
ax1.set_ylabel('Post-Semester GPA')
ax1.set_title('AI Hours vs Post GPA by Major')
ax1.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Box plot of AI usage by major
ax2 = axes[1]
sns.boxplot(data=df_clean, x='Major_Category', y='Post_Semester_GPA',
            hue='AI_Usage_Category', ax=ax2)
ax2.set_xlabel('Major Category')
ax2.set_ylabel('Post-Semester GPA')
ax2.set_title('Post GPA by Major and AI Usage')
ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# %%
# 4.4 Analysis by Primary Use Case
use_case_analysis = df_clean.groupby('Primary_Use_Case').agg({
    'Weekly_GenAI_Hours': 'mean',
    'Post_Semester_GPA': 'mean',
    'Pre_Semester_GPA': 'mean',
    'GPA_Change': 'mean',
    'Skill_Retention_Score': 'mean',
    'Student_ID': 'count'
}).round(2)

use_case_analysis.columns = ['Avg_AI_Hours', 'Avg_Post_GPA', 'Avg_Pre_GPA',
                              'Avg_GPA_Change', 'Avg_Skill_Retention', 'Count']
use_case_analysis = use_case_analysis.sort_values('Avg_GPA_Change', ascending=False)

print("📈 Performance by Primary AI Use Case:")
display(use_case_analysis)

# %%
# 4.5 Impact of Paid Subscription
paid_analysis = df_clean.groupby('Paid_Subscription').agg({
    'Weekly_GenAI_Hours': 'mean',
    'Tool_Diversity': 'mean',
    'Post_Semester_GPA': 'mean',
    'GPA_Change': 'mean',
    'Skill_Retention_Score': 'mean',
    'Student_ID': 'count'
}).round(2)

print("💰 Impact of Paid Subscriptions:")
display(paid_analysis)

# Statistical test
from scipy.stats import ttest_ind
paid_gpa = df_clean[df_clean['Paid_Subscription'] == True]['GPA_Change']
free_gpa = df_clean[df_clean['Paid_Subscription'] == False]['GPA_Change']
t_stat, p_value = ttest_ind(paid_gpa, free_gpa)
print(f"\nT-test for GPA Change (Paid vs Free): p-value = {p_value:.4f}")
print("Significant difference" if p_value < 0.05 else "No significant difference")

# %%
# 4.6 Skill Retention Analysis
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Skill retention by AI usage
ax1 = axes[0]
ai_hours_bins = pd.cut(df_clean['Weekly_GenAI_Hours'], bins=[0, 5, 15, 30, 100],
                        labels=['0-5h', '5-15h', '15-30h', '30h+'])
skill_by_ai = df_clean.groupby(ai_hours_bins)['Skill_Retention_Score'].agg(['mean', 'std'])

ax1.bar(skill_by_ai.index, skill_by_ai['mean'], yerr=skill_by_ai['std'],
        capsize=5, color='lightblue', edgecolor='black')
ax1.set_xlabel('Weekly AI Hours')
ax1.set_ylabel('Skill Retention Score')
ax1.set_title('Skill Retention by AI Usage')

# Skill retention by dependency
ax2 = axes[1]
dep_categories = ['Low', 'Medium', 'High', 'Very High']
skill_by_dep = df_clean.groupby('Dependency_Category')['Skill_Retention_Score'].agg(['mean', 'std'])

ax2.bar(skill_by_dep.index, skill_by_dep['mean'], yerr=skill_by_dep['std'],
        capsize=5, color='lightcoral', edgecolor='black')
ax2.set_xlabel('AI Dependency Level')
ax2.set_ylabel('Skill Retention Score')
ax2.set_title('Skill Retention by AI Dependency')

plt.tight_layout()
plt.show()

# %% [markdown]
# ### 5. Statistical Analysis

# %%
# 5.1 ANOVA Test: Does AI usage differ across majors?
from scipy.stats import f_oneway

major_groups = [df_clean[df_clean['Major_Category'] == major]['Weekly_GenAI_Hours'].values
                for major in df_clean['Major_Category'].unique()]
f_stat, p_value = f_oneway(*major_groups)

print("📊 ANOVA: AI Hours Across Majors")
print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.6f}")
print("Significant difference" if p_value < 0.05 else "No significant difference")

# 5.2 Chi-square test: AI Usage vs Burnout Risk
from scipy.stats import chi2_contingency

contingency = pd.crosstab(df_clean['AI_Usage_Category'], df_clean['Burnout_Risk_Level'])
chi2, p, dof, expected = chi2_contingency(contingency)

print("\n📊 Chi-square: AI Usage vs Burnout Risk")
print(f"Chi-square: {chi2:.4f}")
print(f"P-value: {p:.6f}")
print("Significant association" if p < 0.05 else "No significant association")

# 5.3 Correlation between AI Dependency and Anxiety
correlation, p_value = pearsonr(df_clean['Perceived_AI_Dependency'],
                                 df_clean['Anxiety_Level_During_Exams'])

print("\n📊 Correlation: AI Dependency vs Exam Anxiety")
print(f"Pearson correlation: {correlation:.4f}")
print(f"P-value: {p_value:.6f}")
print("Significant correlation" if p_value < 0.05 else "No significant correlation")

# %% [markdown]
# ### 6. Machine Learning Models

# %%
# 6.1 Prepare features for prediction
# Define features for GPA prediction
feature_cols_gpa = ['Pre_Semester_GPA', 'Weekly_GenAI_Hours', 'Tool_Diversity',
                    'Traditional_Study_Hours', 'Perceived_AI_Dependency',
                    'Anxiety_Level_During_Exams', 'Major_Category_encoded',
                    'Primary_Use_Case_encoded', 'Prompt_Engineering_Skill_encoded',
                    'Institutional_Policy_encoded', 'Paid_Subscription_encoded']

# Prepare the data
X_gpa = df_clean[feature_cols_gpa].copy()
y_gpa = df_clean['Post_Semester_GPA']

# Remove rows with NaN
X_gpa = X_gpa.dropna()
y_gpa = y_gpa[X_gpa.index]

# Split the data
X_train_gpa, X_test_gpa, y_train_gpa, y_test_gpa = train_test_split(
    X_gpa, y_gpa, test_size=0.2, random_state=42
)

# Scale the features
scaler_gpa = StandardScaler()
X_train_gpa_scaled = scaler_gpa.fit_transform(X_train_gpa)
X_test_gpa_scaled = scaler_gpa.transform(X_test_gpa)

print(f"Training set size: {len(X_train_gpa)}")
print(f"Test set size: {len(X_test_gpa)}")
print(f"Features: {feature_cols_gpa}")

# %%
# 6.2 Train and evaluate GPA Prediction Models

# Model 1: Linear Regression
lr_model = LinearRegression()
lr_model.fit(X_train_gpa_scaled, y_train_gpa)
y_pred_lr = lr_model.predict(X_test_gpa_scaled)

# Model 2: Random Forest Regressor
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train_gpa_scaled, y_train_gpa)
y_pred_rf = rf_model.predict(X_test_gpa_scaled)

# Model 3: XGBoost Regressor
xgb_model = XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
xgb_model.fit(X_train_gpa_scaled, y_train_gpa)
y_pred_xgb = xgb_model.predict(X_test_gpa_scaled)

# Evaluate models
print("=" * 60)
print("🎯 GPA PREDICTION MODEL RESULTS")
print("=" * 60)

models = {
    'Linear Regression': (lr_model, y_pred_lr),
    'Random Forest': (rf_model, y_pred_rf),
    'XGBoost': (xgb_model, y_pred_xgb)
}

results = {}
for name, (model, predictions) in models.items():
    mse = mean_squared_error(y_test_gpa, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test_gpa, predictions)
    results[name] = {'RMSE': rmse, 'R²': r2}
    print(f"\n{name}:")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  R² Score: {r2:.4f}")

# Feature importance for Random Forest
feature_importance = pd.DataFrame({
    'Feature': feature_cols_gpa,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=False)

print("\n📊 Top 5 Most Important Features for GPA Prediction:")
print(feature_importance.head(5))

# %%
# 6.3 Burnout Risk Classification

# Prepare features for classification
feature_cols_burnout = ['Pre_Semester_GPA', 'Weekly_GenAI_Hours', 'Tool_Diversity',
                        'Traditional_Study_Hours', 'Perceived_AI_Dependency',
                        'Anxiety_Level_During_Exams', 'Skill_Retention_Score',
                        'Major_Category_encoded', 'Primary_Use_Case_encoded',
                        'Institutional_Policy_encoded']

X_burnout = df_clean[feature_cols_burnout].copy()
y_burnout = df_clean['Burnout_Risk_Level_encoded']

# Remove NaN
X_burnout = X_burnout.dropna()
y_burnout = y_burnout[X_burnout.index]

# Split data
X_train_brn, X_test_brn, y_train_brn, y_test_brn = train_test_split(
    X_burnout, y_burnout, test_size=0.2, random_state=42, stratify=y_burnout
)

# Scale features
scaler_brn = StandardScaler()
X_train_brn_scaled = scaler_brn.fit_transform(X_train_brn)
X_test_brn_scaled = scaler_brn.transform(X_test_brn)

# Handle class imbalance with SMOTE
smote = SMOTE(random_state=42)
X_train_brn_resampled, y_train_brn_resampled = smote.fit_resample(X_train_brn_scaled, y_train_brn)

print(f"Original class distribution: {np.bincount(y_train_brn)}")
print(f"Resampled class distribution: {np.bincount(y_train_brn_resampled)}")

# %%
# 6.4 Train Burnout Classification Models

# Model 1: Random Forest Classifier
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train_brn_resampled, y_train_brn_resampled)
y_pred_rf_clf = rf_clf.predict(X_test_brn_scaled)

# Model 2: Gradient Boosting Classifier
gb_clf = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb_clf.fit(X_train_brn_resampled, y_train_brn_resampled)
y_pred_gb_clf = gb_clf.predict(X_test_brn_scaled)

# Model 3: XGBoost Classifier
xgb_clf = XGBClassifier(n_estimators=100, random_state=42, eval_metric='mlogloss')
xgb_clf.fit(X_train_brn_resampled, y_train_brn_resampled)
y_pred_xgb_clf = xgb_clf.predict(X_test_brn_scaled)

print("=" * 60)
print("🔥 BURNOUT RISK CLASSIFICATION RESULTS")
print("=" * 60)

classification_results = {}
for name, (model, predictions) in zip(['Random Forest', 'Gradient Boosting', 'XGBoost'],
                                        [(rf_clf, y_pred_rf_clf), (gb_clf, y_pred_gb_clf),
                                         (xgb_clf, y_pred_xgb_clf)]):
    accuracy = accuracy_score(y_test_brn, predictions)
    classification_results[name] = {'Accuracy': accuracy}
    print(f"\n{name}:")
    print(f"  Accuracy: {accuracy:.4f}")
    print("\n  Classification Report:")
    print(classification_report(y_test_brn, predictions,
                                 target_names=['Low', 'Medium', 'High']))

# Best model feature importance
best_clf = xgb_clf if classification_results['XGBoost']['Accuracy'] > max(
    classification_results['Random Forest']['Accuracy'],
    classification_results['Gradient Boosting']['Accuracy']) else rf_clf

if best_clf == xgb_clf:
    importance = best_clf.feature_importances_
else:
    importance = best_clf.feature_importances_

feature_importance_burnout = pd.DataFrame({
    'Feature': feature_cols_burnout,
    'Importance': importance
}).sort_values('Importance', ascending=False)

print("\n📊 Top Features for Burnout Prediction:")
print(feature_importance_burnout.head(5))

# %% [markdown]
# ### 7. Clustering Analysis: Student Segmentation

# %%
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# Select features for clustering
cluster_features = ['Weekly_GenAI_Hours', 'Traditional_Study_Hours',
                    'Perceived_AI_Dependency', 'Skill_Retention_Score',
                    'Post_Semester_GPA']

X_cluster = df_clean[cluster_features].dropna()

# Scale the data
scaler_cluster = StandardScaler()
X_cluster_scaled = scaler_cluster.fit_transform(X_cluster)

# Find optimal number of clusters using Elbow method
inertias = []
silhouette_scores = []
K_range = range(2, 10)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_cluster_scaled)
    inertias.append(kmeans.inertia_)

    from sklearn.metrics import silhouette_score
    silhouette_scores.append(silhouette_score(X_cluster_scaled, kmeans.labels_))

# Plot elbow curve
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(K_range, inertias, 'bo-')
axes[0].set_xlabel('Number of clusters (k)')
axes[0].set_ylabel('Inertia')
axes[0].set_title('Elbow Method for Optimal k')

axes[1].plot(K_range, silhouette_scores, 'ro-')
axes[1].set_xlabel('Number of clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Score for Optimal k')

plt.tight_layout()
plt.show()

# Choose k=4 for clustering
kmeans_final = KMeans(n_clusters=4, random_state=42, n_init=10)
clusters = kmeans_final.fit_predict(X_cluster_scaled)

# Add cluster labels to dataframe
df_clustered = df_clean.iloc[X_cluster.index].copy()
df_clustered['Cluster'] = clusters

# Visualize clusters with PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_cluster_scaled)

plt.figure(figsize=(12, 8))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='viridis', alpha=0.6)
plt.colorbar(scatter, label='Cluster')
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.title('Student Clusters Visualization (PCA)')
plt.show()

# Analyze cluster characteristics
print("=" * 60)
print("👥 STUDENT CLUSTER ANALYSIS")
print("=" * 60)

cluster_analysis = df_clustered.groupby('Cluster').agg({
    'Weekly_GenAI_Hours': 'mean',
    'Traditional_Study_Hours': 'mean',
    'Perceived_AI_Dependency': 'mean',
    'Skill_Retention_Score': 'mean',
    'Post_Semester_GPA': 'mean',
    'Student_ID': 'count'
}).round(2)

cluster_analysis.columns = ['Avg_AI_Hours', 'Avg_Study_Hours', 'Avg_Dependency',
                            'Avg_Skill_Retention', 'Avg_GPA', 'Count']
print("\nCluster Profiles:")
display(cluster_analysis)

# Name the clusters based on characteristics
cluster_names = {
    0: "📚 Traditional Learners",
    1: "🦾 AI Power Users",
    2: "⚖️ Balanced Learners",
    3: "⚠️ High-Risk Users"
}

for cluster in range(4):
    print(f"\n{cluster_names[cluster]}:")
    print(f"  Size: {cluster_analysis.loc[cluster, 'Count']} students")
    print(f"  Avg AI Hours: {cluster_analysis.loc[cluster, 'Avg_AI_Hours']:.1f}h/week")
    print(f"  Avg GPA: {cluster_analysis.loc[cluster, 'Avg_GPA']:.2f}")
    print(f"  Avg Skill Retention: {cluster_analysis.loc[cluster, 'Avg_Skill_Retention']:.1f}%")

# %% [markdown]
# ### 8. Interactive Analysis Dashboard

# %%
# 8.1 Optimal AI Usage Analysis
def find_optimal_ai_hours(df):
    """Find the AI usage range associated with highest GPA"""
    df_temp = df.copy()
    df_temp['AI_Bucket'] = pd.cut(df_temp['Weekly_GenAI_Hours'],
                                   bins=[0, 2, 5, 10, 15, 20, 30, 50])
    optimal = df_temp.groupby('AI_Bucket')['Post_Semester_GPA'].agg(['mean', 'count'])
    optimal = optimal[optimal['count'] > 10]
    best_bucket = optimal['mean'].idxmax()
    return best_bucket, optimal

optimal_bucket, optimal_data = find_optimal_ai_hours(df_clean)

plt.figure(figsize=(12, 6))
plt.bar(range(len(optimal_data)), optimal_data['mean'],
        color='skyblue', edgecolor='black')
plt.xticks(range(len(optimal_data)), [str(b) for b in optimal_data.index], rotation=45)
plt.xlabel('Weekly AI Hours Range')
plt.ylabel('Average Post-Semester GPA')
plt.title(f'Optimal AI Usage: {optimal_bucket} hours/week')
plt.tight_layout()
plt.show()

print(f"\n✨ OPTIMAL AI USAGE: {optimal_bucket} hours/week")
print(f"   Average GPA in this range: {optimal_data.loc[optimal_bucket, 'mean']:.3f}")

# %%
# 8.2 Recommendations by Major
recommendations = df_clean.groupby('Major_Category').agg({
    'Weekly_GenAI_Hours': ['mean', 'median'],
    'Post_Semester_GPA': 'mean',
    'Tool_Diversity': 'mean',
    'Perceived_AI_Dependency': 'mean'
}).round(2)

recommendations.columns = ['Avg_AI_Hours', 'Median_AI_Hours', 'Avg_GPA', 'Avg_Tools', 'Avg_Dependency']
recommendations = recommendations.sort_values('Avg_GPA', ascending=False)

print("=" * 60)
print("💡 MAJOR-SPECIFIC RECOMMENDATIONS")
print("=" * 60)
display(recommendations)

# Generate recommendations
for major in recommendations.index:
    avg_ai = recommendations.loc[major, 'Avg_AI_Hours']
    avg_gpa = recommendations.loc[major, 'Avg_GPA']
    avg_dep = recommendations.loc[major, 'Avg_Dependency']

    print(f"\n📌 {major}:")
    if avg_ai < 5:
        print(f"   → Consider increasing AI usage (currently {avg_ai:.1f}h/week)")
    elif avg_ai > 20:
        print(f"   → Reduce AI usage to prevent burnout ({avg_ai:.1f}h/week)")
    else:
        print(f"   → Current AI usage ({avg_ai:.1f}h/week) seems balanced")

    if avg_dep > 6:
        print(f"   → High AI dependency detected ({avg_dep:.1f}/10). Focus on skill development")
    else:
        print(f"   → Healthy AI dependency level ({avg_dep:.1f}/10)")

# %% [markdown]
# ### 9. Key Findings and Recommendations

# %%
# Generate comprehensive report
print("=" * 80)
print("📊 AI IN EDUCATION: COMPREHENSIVE ANALYSIS REPORT")
print("=" * 80)

print("\n🔍 KEY FINDINGS:")

# Finding 1: AI usage distribution
print(f"\n1. AI ADOPTION:")
print(f"   - Average weekly AI usage: {df_clean['Weekly_GenAI_Hours'].mean():.1f} hours")
print(f"   - {df_clean[df_clean['Weekly_GenAI_Hours'] > 20].shape[0]} students use AI >20h/week")
print(f"   - Most common use case: {df_clean['Primary_Use_Case'].mode()[0]}")

# Finding 2: Impact on GPA
low_ai = df_clean[df_clean['Weekly_GenAI_Hours'] <= 5]['GPA_Change'].mean()
med_ai = df_clean[(df_clean['Weekly_GenAI_Hours'] > 5) & (df_clean['Weekly_GenAI_Hours'] <= 15)]['GPA_Change'].mean()
high_ai = df_clean[df_clean['Weekly_GenAI_Hours'] > 15]['GPA_Change'].mean()

print(f"\n2. GPA IMPACT BY AI USAGE:")
print(f"   - Low AI users (≤5h): GPA Change = {low_ai:.3f}")
print(f"   - Moderate AI users (5-15h): GPA Change = {med_ai:.3f}")
print(f"   - High AI users (>15h): GPA Change = {high_ai:.3f}")

# Finding 3: Burnout correlation
burnout_by_ai = df_clean.groupby('AI_Usage_Category')['Burnout_Risk_Level'].value_counts(normalize=True).unstack()
print(f"\n3. BURNOUT RISK BY AI USAGE:")
for category in burnout_by_ai.index:
    high_pct = burnout_by_ai.loc[category, 'High'] * 100 if 'High' in burnout_by_ai.columns else 0
    print(f"   - {category}: {high_pct:.1f}% High Burnout Risk")

# Finding 4: Best practices
print(f"\n4. BEST PRACTICES OBSERVED:")
print(f"   - Optimal AI usage: {optimal_bucket} hours/week")
print(f"   - Tool diversity average: {df_clean['Tool_Diversity'].mean():.1f} tools")
print(f"   - Paid subscription users: {df_clean['Paid_Subscription'].sum()} ({df_clean['Paid_Subscription'].mean()*100:.1f}%)")

print("\n" + "=" * 80)
print("📋 ACTIONABLE RECOMMENDATIONS")
print("=" * 80)

print("\nFOR STUDENTS:")
print("1. ⏰ Limit AI usage to 5-15 hours per week for optimal GPA improvement")
print("2. 🚠 Use 3-4 different AI tools to diversify learning approaches")
print("3. 📚 Maintain traditional study habits alongside AI tools")
print("4. 🎯 Focus AI usage on summarization and debugging rather than direct answer generation")
print("5. 🔄 Practice prompt engineering skills to get better quality outputs")

print("\nFOR EDUCATIONAL INSTITUTIONS:")
print("1. 📋 Develop AI literacy programs teaching effective prompt engineering")
print("2. 🃥 Provide mental health support for high-AI dependency students")
print("3. 🎓 Implement major-specific AI guidelines")
print("4. 💰 Consider subsidizing premium AI tools for students")
print("5. 📊 Monitor AI usage patterns to identify at-risk students early")

print("\nFOR POLICY MAKERS:")
print("1. ⚖️ Create balanced AI policies (neither strict ban nor complete allowance)")
print("2. 🧠 Integrate AI ethics and critical thinking in curriculum")
print("3. 📈 Fund research on long-term AI impact on learning outcomes")

# %% [markdown]
# ### 10. Save Results and Export

# %%
# Save analysis results
from datetime import datetime

# Create results DataFrame
results_summary = pd.DataFrame({
    'Metric': ['Total Students', 'Average Weekly AI Hours', 'Average GPA',
               'Average Skill Retention', 'High Burnout Rate', 'Optimal AI Hours'],
    'Value': [len(df), f"{df_clean['Weekly_GenAI_Hours'].mean():.1f}",
              f"{df_clean['Post_Semester_GPA'].mean():.2f}",
              f"{df_clean['Skill_Retention_Score'].mean():.1f}%",
              f"{(df_clean['Burnout_Risk_Level'] == 'High').mean()*100:.1f}%",
              f"{optimal_bucket}"]
})

print("🗂️ Saving analysis results...")
results_summary.to_csv('ai_impact_summary.csv', index=False)
print("✅ Results saved to 'ai_impact_summary.csv'")

# Save cluster analysis
cluster_analysis.to_csv('student_clusters.csv')
print("✅ Cluster data saved to 'student_clusters.csv'")

# Save recommendations
recommendations.to_csv('major_recommendations.csv')
print("✅ Major recommendations saved to 'major_recommendations.csv'")

print("\n🎉 PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
