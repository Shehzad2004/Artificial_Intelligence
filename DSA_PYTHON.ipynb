Find Sum and Average of List

Concept: basic operations

```python
nums = [10, 20, 30, 40]

total = sum(nums)
average = total / len(nums)

print("Sum:", total)
print("Average:", average)
```

Remove Duplicates from List

Concept: list + set

```python
nums = [1, 2, 2, 3, 4, 4, 5]

unique = list(set(nums))
print(unique)
```

Reverse a List Without Built-in Function

Concept: algorithmic thinking

```python
nums = [1, 2, 3, 4]
reversed_list = []

for i in range(len(nums)-1, -1, -1):
    reversed_list.append(nums[i])

print(reversed_list)
```

Linear Search

Concept: searching algorithm

```python
nums = [5, 8, 12, 20]
target = 12

for i in range(len(nums)):
    if nums[i] == target:
        print("Found at index", i)
        break
```

Sorting a List (Bubble Sort)

Concept: sorting algorithm

```python
nums = [5, 1, 4, 2]

for i in range(len(nums)):
    for j in range(0, len(nums)-i-1):
        if nums[j] > nums[j+1]:
            nums[j], nums[j+1] = nums[j+1], nums[j]

print(nums)
```

Merge Two Sorted Lists

Concept: two-pointer technique

```python
a = [1, 3, 5]
b = [2, 4, 6]

merged = []
i = j = 0

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        merged.append(a[i])
        i += 1
    else:
        merged.append(b[j])
        j += 1

merged.extend(a[i:])
merged.extend(b[j:])

print(merged)
```

Find Subarray with Maximum Sum (Kadane’s Algorithm)

Concept: optimized algorithm (DSA level)

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = current_sum = nums[0]

for num in nums[1:]:
    current_sum = max(num, current_sum + num)
    max_sum = max(max_sum, current_sum)

print("Maximum Subarray Sum:", max_sum)

```
